# 🏆 2026 FIFA World Cup Knockout Stage Analysis & Predictions
This notebook provides a comprehensive analysis of the final stages of the 2026 FIFA World Cup. It includes the remaining schedule, team performance metrics, predictive modeling for matchups, and visual tracking of the tournament tree.

### 1. Setup and Environment Initialization

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import networkx as nx

# Set professional design style
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["figure.figsize"] = (10, 6)
```

### 2. Golden Boot Race Analysis
Currently, Kylian Mbappé and Lionel Messi are leading the tournament tracking with 8 goals each heading into the deep knockout phase.

```python
scorers_data = {
    'Player': ['Kylian Mbappé', 'Lionel Messi', 'Erling Haaland', 'Olivier Giroud', 'Bukayo Saka'],
    'Team': ['France', 'Argentina', 'Norway', 'France', 'England'],
    'Goals': [8, 8, 5, 4, 4]
}
df_scorers = pd.DataFrame(scorers_data)

plt.figure(figsize=(10, 5))
colors = ['#1d3557', '#457b9d', '#e63946', '#a8dadc', '#f1faee']
sns.barplot(x='Goals', y='Player', data=df_scorers, palette=colors, hue='Player', legend=False)
plt.title('2026 World Cup Top Scorers (Quarter-Finals Stage)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Goals Scored', fontsize=12)
plt.ylabel('Player', fontsize=12)
for index, value in enumerate(df_scorers['Goals']):
    plt.text(value + 0.1, index, str(value), va='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 3. Tournament Bracket Visualization
Let's visualize the knockout progression from the Quarter-Finals up to the Grand Final.

```python
G = nx.DiGraph()

# Add nodes with positions to form a bracket hierarchy
pos = {
    'France': (0, 3), 'Morocco': (0, 2), 'Spain': (0, 1), 'Belgium': (0, 0),
    'Norway': (0, -1), 'England': (0, -2), 'Argentina': (0, -3), 'Switzerland': (0, -4),
    'Semi 1 A': (2, 2.5), 'Semi 1 B': (2, 0.5),
    'Semi 2 A': (2, -1.5), 'Semi 2 B': (2, -3.5),
    'Finalist 1': (4, 1.5), 'Finalist 2': (4, -2.5),
    'Champion': (6, -0.5)
}

edges = [
    ('France', 'Semi 1 A'), ('Morocco', 'Semi 1 A'),
    ('Spain', 'Semi 1 B'), ('Belgium', 'Semi 1 B'),
    ('Norway', 'Semi 2 A'), ('England', 'Semi 2 A'),
    ('Argentina', 'Semi 2 B'), ('Switzerland', 'Semi 2 B'),
    ('Semi 1 A', 'Finalist 1'), ('Semi 1 B', 'Finalist 1'),
    ('Semi 2 A', 'Finalist 2'), ('Semi 2 B', 'Finalist 2'),
    ('Finalist 1', 'Champion'), ('Finalist 2', 'Champion')
]
G.add_edges_from(edges)

plt.figure(figsize=(12, 8))
nx.draw(G, pos, with_labels=True, node_color='#eaeaea', node_size=2500, font_size=9, font_weight='bold', edge_color='#b0c4de', arrows=True)
plt.title('2026 World Cup Bracket Architecture Map', fontsize=14, fontweight='bold')
plt.show()
```
